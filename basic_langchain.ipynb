### Simple QnA Bot without LangChain - With Groq


```python

from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": "hi"
      },
      {
        "role": "assistant",
        "content": "Explain me about groq",
      },
      {
        "role": "user",
        "content": ""
      }
    ],
    temperature=1,
    max_completion_tokens=2048,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)
for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
    
```

### `Promt Templates` with LangChain

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

MODEL = "groq:openai/gpt-oss-120b"

llm = init_chat_model(
    model=MODEL,
    temperature=1,
    max_completion_tokens=2048,
    top_p=1,
    reasoning_effort="medium",
)

prompt_template = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a Python developer with 7 years of experience."
    ),
    MessagesPlaceholder(variable_name="user_input")
])

prompt = prompt_template.invoke({
    "user_input": [
        HumanMessage(content="What is a Python decorator?")
    ]
})

response = llm.invoke(prompt)

print(response.content)
```

### 3. LCEL - | Operator


```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 1. Model
MODEL = "groq:openai/gpt-oss-120b"

llm = init_chat_model(
    model=MODEL,
    temperature=1,
    max_completion_tokens=2048,
    top_p=1,
    reasoning_effort="medium",
)

# 2. Prompt Template
prompt_template = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a Python developer with 7 years of experience."
    ),
    MessagesPlaceholder(variable_name="user_input")
])

# 3. LCEL
chain = prompt_template | llm

# 4. Run the chain
response = chain.invoke({
    "user_input": [
        HumanMessage(content="Explain Python decorator in simple words.")
    ]
})

# 5. Output
print(response.content)
```

### 4. Memory - Conver. Histroy

```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# 1. Model
MODEL = "groq:openai/gpt-oss-120b"

llm = init_chat_model(
    model=MODEL,
    temperature=1,
    max_completion_tokens=2048,
    reasoning_effort="medium",
)

# 2. Prompt
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),

    # Short-term memory এখানে যাবে
    MessagesPlaceholder(variable_name="chat_history"),

    ("human", "{user_input}")
])

# 3. LCEL
chain = prompt_template | llm

# 4. Short-term memory
chat_history = []

# First question
response = chain.invoke({
    "chat_history": chat_history,
    "user_input": "My favorite programming language is Python."
})

print(response.content)

# Memory-তে conversation যোগ করি
chat_history.append(
    HumanMessage(content="My favorite programming language is Python.")
)

chat_history.append(
    AIMessage(content=response.content)
)

# Second question
response = chain.invoke({
    "chat_history": chat_history,
    "user_input": "What is my favorite programming language?"
})

print(response.content)
```

### 5. Structured output



```python
from pydantic import BaseModel
from langchain.chat_models import init_chat_model


# 1. Structure / Schema
class Person(BaseModel):
    name: str
    age: int
    favorite_language: str


# 2. LLM
llm = init_chat_model(
    model="groq:openai/gpt-oss-120b"
)


# 3. Structured LLM
structured_llm = llm.with_structured_output(Person)


# 4. Invoke
response = structured_llm.invoke(
    "Mehedi is 30 years old and his favorite programming language is Python."
)

print(response)
```
